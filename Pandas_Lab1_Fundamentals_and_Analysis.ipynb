- `pd.DataFrame` → 2D table, many columns, each column is a Series

## 10. `.loc[]` vs `.iloc[]`

- `.loc[row_label, col_label]` → label based
- `.iloc[row_pos, col_pos]` → position based (integers, like list indexing)

```python
df.loc[0, "name"]
df.iloc[0, 1]
df.loc[0:2, ["name", "age"]]
df.iloc[0:2, 0:2]
```

## 12. Common Indexing Syntax Errors (read & understand)

- `df[df.age > 20 and df.marks > 70]`  ❌ → use `&` not `and`
- `df["age" > 20]`  ❌ → condition must be on the column, not the string
- `df.loc[0, 1]`  ❌ → `.loc` doesn't take integer positions for columns unless the column labels ARE integers
